# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

````python
# 🛠️ TOOL 3 (BONUS): Word Counter

def word_count(text: str) -> int:
    """Count the number of words in a piece of text."""
    try:
        return len(text.split())
    except Exception:
        return 0
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
# 🤖 AGENT FUNCTION (TO IMPLEMENT)

def agent(query: str):
    query_lower = query.lower()

    # 📝 Bonus: simple logging of every incoming query
    print(f"[LOG] Received query: {query}")

    try:
        # Routing logic
        if "calculate" in query_lower:
            expression = query_lower.replace("calculate", "").strip()
            result = calculator(expression)
            if result == "Error in calculation":
                print("[LOG] Calculator tool returned an error")
                return {"type": "error", "result": result}
            print("[LOG] Routed to Calculator Tool")
            return {"type": "calculation", "result": result}

        elif "keywords" in query_lower:
            keywords = extract_keywords(query)
            print("[LOG] Routed to Keyword Extractor Tool")
            return {"type": "keywords", "result": keywords}

        # extra routing for the word counter tool
        elif "count words" in query_lower or "word count" in query_lower:
            text = query_lower.replace("count words", "").replace("word count", "").strip()
            count = word_count(text)
            print("[LOG] Routed to Word Counter Tool")
            return {"type": "word_count", "result": count}

        else:
            print("[LOG] Routed to General Response")
            return {
                "type": "general",
                "result": f"I received your query: '{query}'. I can help with calculations, keyword extraction, or word counts."
            }

    except Exception as e:
        # Basic error handling
        print(f"[ERROR] {e}")
        return {"type": "error", "result": str(e)}
````

## 📦 Expected Output Format

```
{
  "type": "calculation / keywords / general / error",
  "result": ...
}
```

````python
# 🧪 Test Cases

queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?"
]

for q in queries:
    print("Query:", q)
    print("Response:", agent(q))
    print("-" * 50)
````

````python
# 🎯 Interactive Mode

while True:
    user_input = input("Enter query (type 'exit' to stop): ")
    if user_input.lower() == "exit":
        break
    print("Response:", agent(user_input))
````
